# 1- Temos uma quantidade equilibrada de exemplos comestíveis e venenosos para analisar?

Qual a proporção de cogumelos comestíveis para os não comestíveis?

Quantos cogumelos são classificados como comestíveis e quantos como venenosos?

```python
# Plotando o gráfico de barras
plt.figure(figsize=(6, 4))
ax = sns.countplot(data=df, x='class', hue='class', palette={'e': '#2ecc71', 'p': '#e74c3c'}, legend=False)

# Adicionando os valores em cima de cada coluna
for container in ax.containers:
    ax.bar_label(container, color='black')

plt.title('Proporção: Comestíveis (e) vs Venenosos (p)')
plt.xlabel('Classe do Cogumelo')
plt.ylabel('Quantidade de Amostras')
ax.set_xticklabels(["Venenoso", "Comestível"])
plt.show()
```

```python
contagem = df["class"].value_counts()

plt.figure(figsize=(6, 6))

plt.pie(
    contagem,
    labels=["Venenoso", "Comestível"],
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Proporção de Cogumelos por Classe")
plt.show()
```

# 2- Dentro de cada tipo de odor, qual é a proporção de cogumelos comestíveis e venenosos?

```python
proporcao = pd.crosstab(
    df["odor"],
    df["class"],
    normalize="index"
) * 100

proporcao.plot(
    kind="bar",
    stacked=True,
    figsize=(10, 5)
)

plt.title("Proporção das Classes por Odor")
plt.xlabel("Odor")
plt.ylabel("Proporção (%)")
plt.legend(title="Classe")
plt.xticks(rotation=0)
plt.show()
```

# 3- O cheiro do cogumelo é suficiente para determinar se ele vai me matar?

```python
plt.figure(figsize=(10, 6))
# Usando 'hue' para separar comestíveis e venenosos dentro de cada odor
sns.countplot(data=df, x='odor', hue='class', palette={'e': '#2ecc71', 'p': '#e74c3c'})
plt.title('Relação entre o Odor e a Toxicidade')
plt.xlabel('Tipo de Odor (n=nenhum, f=fétido, etc.)')
plt.ylabel('Quantidade')
plt.legend(title='Classe', labels=['Comestível', 'Venenoso'])
plt.show()
```

# 4- Quais combinações exatas de cor de esporo garantem que o cogumelo é venenoso?

```python
plt.figure(figsize=(8, 5))
# Criando uma matriz de contagem entre as duas colunas
crosstab = pd.crosstab(df['spore-print-color'], df['class'])

# Plotando o heatmap com os números (annot=True)
sns.heatmap(crosstab, annot=True, fmt='d', cmap='Reds', cbar=True)
plt.title('Concentração: Cor da Impressão de Esporos vs Classe')
plt.xlabel('Cor da Impressão de Esporos')
plt.ylabel('Classe (e = Comestível, p = Venenoso)')
plt.show()
```

# 5- Como os registros de cada tipo de odor estão distribuídos entre as classes?

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x="bruises", hue="class")

plt.title("Distribuição da Classe por Presença de Hematomas")
plt.xlabel("Hematomas")
plt.ylabel("Quantidade")
plt.show()
```

# 6- Quais tipos de odor são mais frequentes entre os cogumelos?

# 7- Quais cores de chapéu aparecem com maior frequência?

```python
plt.figure(figsize=(10, 5))

sns.countplot(
    data=df,
    x="cap-color",
    order=df["cap-color"].value_counts().index
)

plt.title("Distribuição das Cores do Chapéu")
plt.xlabel("Cor do chapéu")
plt.ylabel("Quantidade")
plt.show()
```

# 8- Existe uma diferença na distribuição entre cogumelos comestíveis e venenosos de acordo com o odor?

```python
plt.figure(figsize=(10, 5))

sns.countplot(data=df, x="odor", hue="class")

plt.title("Distribuição da Classe por Odor")
plt.xlabel("Odor")
plt.ylabel("Quantidade")
plt.show()
```

# 9- A distribuição de cogumelos comestíveis e venenosos varia de acordo com o habitat?
