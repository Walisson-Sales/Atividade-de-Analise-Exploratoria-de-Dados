### **1- Qual a proporção de cogumelos comestíveis para os não comestíveis?**

```python
# Dados
contagem = df["class"].value_counts()
valores = [contagem['e'], contagem['p']]
rotulos = ["Comestível", "Venenoso"]
cores = ['#2ecc71', '#e74c3c']

# Criando a figura com 1 linha e 2 colunas
fig, ax = plt.subplots(1, 2, figsize=(12, 5))

# --- GRÁFICO 1: BARRAS (ax[0]) ---
sns.barplot(
    x=rotulos,
    y=valores,
    hue=rotulos,
    palette=cores,
    ax=ax[0],
    legend=False
)
ax[0].set_title("Valores Absolutos")
ax[0].set_ylabel("Quantidade")

# Adicionando os números exatos no topo das barras
for i, v in enumerate(valores):
    ax[0].text(i, v + 50, str(v), ha='center', fontweight='bold')

# --- GRÁFICO 2: PIZZA (ax[1]) ---
ax[1].pie(
    valores,
    labels=rotulos,
    colors=cores,
    autopct="%1.1f%%",
    startangle=90,
    explode=(0.05, 0),
    shadow=True,
    textprops={'fontsize': 11, 'weight': 'bold'}
)
ax[1].set_title("Proporção Relativa (%)")

# Ajustes finais da figura
plt.suptitle("Análise da Variável Alvo: Classes de Cogumelos", fontsize=14, weight='bold')
plt.tight_layout()
plt.show()
```

#### Conclusão:
Temos registrado mais cogumelo comestíveis do que venenosos, porém é uma diferença mínima de apenas 3,6% (292 cogumelos comestíveis de diferença)

### **2- Dentro de cada tipo de odor, qual é a proporção de cogumelos comestíveis e venenosos?**

```python
# Dicionário para traduzir as letras do odor
mapa_odor = {
    'a': 'Amêndoa', 'l': 'Anis', 'c': 'Creosoto', 'y': 'Peixe',
    'f': 'Fétido', 'm': 'Mofado', 'n': 'Nenhum', 'p': 'Pungente', 's': 'Picante'
}

# Cria uma cópia temporária para o gráfico ou aplica direto no df
df_plot = df.copy()
df_plot['odor_legivel'] = df_plot['odor'].map(mapa_odor)

plt.figure(figsize=(12, 6)) # Um pouco mais largo para acomodar os nomes inteiros

# O seu plot apontando para a coluna legível
ax = sns.countplot(
    data=df_plot,
    x='odor_legivel',
    hue='class',
    palette={'e': '#e74c3c', 'p': '#2ecc71'},
    order=df_plot['odor_legivel'].value_counts().index
)

# Laço de repetição para colocar os números no topo das barras
for container in ax.containers:
    ax.bar_label(container, fmt='%d', padding=3, fontweight='bold')

plt.title('Relação entre o Odor e a Toxicidade', fontsize=14, weight='bold')
plt.xlabel('Tipo de Odor', fontsize=11)
plt.ylabel('Quantidade', fontsize=11)
plt.legend(title='Classe', labels=['Comestível', 'Venenoso'])

# Remove a borda superior e a da direita
sns.despine()

plt.tight_layout()
plt.show()
```

#### Conclusão
O odor é um preditor definitivo de toxicidade, já que, com exceção dos cogumelos sem cheiro ("Nenhum"), todos os outros odores são predominantes de apenas uma classe

### **3- Quais combinações exatas de cor de esporo garantem que o cogumelo é venenoso?**

```python
# Dicionários de Tradução
mapa_esporos = {
    'k': 'Preto', 'n': 'Marrom', 'b': 'Bege', 'h': 'Chocolate',
    'r': 'Verde', 'o': 'Laranja', 'u': 'Roxo', 'w': 'Branco', 'y': 'Amarelo'
}
mapa_classe = {'e': 'Comestível', 'p': 'Venenoso'}

# Cria cópia e aplica traduções
df_plot = df.copy()
df_plot['spore-print-color_legivel'] = df_plot['spore-print-color'].map(mapa_esporos)
df_plot['class_legivel'] = df_plot['class'].map(mapa_classe)

# 2. Criando a matriz de contagem (Y = Esporos, X = Classe)
crosstab = pd.crosstab(df_plot['spore-print-color_legivel'], df_plot['class_legivel'])

plt.figure(figsize=(10, 6))

# 3. Plotando o heatmap com melhorias visuais
sns.heatmap(
    crosstab,
    annot=True,
    fmt='d',
    cmap='Reds',
    linewidths=1.5,
    cbar_kws={'label': 'Quantidade de Amostras'}
)

# Ajustando títulos e eixos (agora nos lugares certos)
plt.title('Concentração: Cor da Impressão de Esporos vs Toxicidade', fontsize=14, weight='bold', pad=15)
plt.xlabel('Classe de Toxicidade', fontsize=11)
plt.ylabel('Cor da Impressão de Esporos', fontsize=11)

# Mantém os nomes das cores na horizontal para leitura imediata
plt.yticks(rotation=0)

plt.tight_layout()
plt.show()
```

#### Conclusão
A única cor de impressão de esporos que é possível garantir que o cogumelo é venenoso é a cor Verde, nas demais não é possível dizer com 100% de precisão

### **4- Como os registros de cada tipo de hematoma estão distribuídos entre as classes?**

```python
plt.figure(figsize=(8, 5))

# Traduzindo os dados numa cópia temporária
df_plot = df.copy()
df_plot['class'] = df_plot['class'].map({'e': 'Comestível', 'p': 'Venenoso'})
df_plot['bruises'] = df_plot['bruises'].map({'t': 'Sim (t)', 'f': 'Não (f)'})

# Plotando de forma direta
sns.countplot(
    data=df_plot,
    x="bruises",
    hue="class",
    palette={'Comestível': '#2ecc71', 'Venenoso': '#e74c3c'},
    order=['Sim (t)', 'Não (f)'],
    hue_order=['Comestível', 'Venenoso']
)

plt.title("Distribuição da Classe por Presença de Hematomas", fontsize=13, weight='bold')
plt.xlabel("Possui Hematomas?")
plt.ylabel("Quantidade")

plt.show()
```

#### Conclusão
A maior parte dos cogumelos com hematomas é comestível, enquanto a grande maioria dos que não possuem hematomas é venenosa.

### **5- Quais combinações de ambiente e agrupamento são fortes indicadores de que ele é seguro ou venenoso?**

```python
# 1. Tratamento e Mapeamento para Leitura Humana
map_habitat = {'g':'Gramíneas', 'l':'Folhas', 'm':'Prados', 'p':'Caminhos', 'u':'Urbano', 'w':'Resíduos', 'd':'Bosques'}
map_populacao = {'a':'Abundante', 'c':'Aglomerada', 'n':'Numerosa', 's':'Espalhada', 'v':'Vários', 'y':'Solitária'}
map_classe = {'e':'Comestível', 'p':'Venenoso'}

df['habitat'] = df['habitat'].map(map_habitat)
df['population'] = df['population'].map(map_populacao)
df['class'] = df['class'].map(map_classe)

# 2. O seu código de plotagem original (agora legível)
g = sns.catplot(
    data=df,
    x="habitat",
    hue="population",
    col="class",
    kind="count",
    height=5,
    aspect=1.2 # Um pouco mais largo para acomodar os nomes inteiros
)

g.set_axis_labels("Habitat", "Quantidade")
g.set_titles("Classe: {col_name}")
plt.show()
```

#### Conclusão
Populações "Abundantes" ou "Numerosas" (especialmente em gramíneas) e habitats de "Resíduos" são fortes indicadores de cogumelos seguros, enquanto agrupamentos de "Vários" encontrados em "Caminhos" ou "Folhas" são alertas fortíssimos de toxicidade.

### **6- A distribuição de cogumelos comestíveis e venenosos varia de acordo com o habitat?**

#### Conclusão

A cor das lâminas é um excelente divisor de classes, onde a cor "b" aponta quase com 100% de certeza para cogumelos venenosos, enquanto cores como "n" e "w" são majoritariamente comestíveis, e as cores "e" e "o" representam espécies totalmente seguras.

```python
# Carregando os dados
df = pd.read_csv('mushrooms.csv')

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Escolhendo 3 colunas categóricas para os eixos X, Y e Z
x_col = 'odor'
y_col = 'stalk-root'
z_col = 'population'

# Convertendo as letras em códigos numéricos (0, 1, 2...)
x_num = df[x_col].astype('category').cat.codes
y_num = df[y_col].astype('category').cat.codes
z_num = df[z_col].astype('category').cat.codes

# Criando o 'Jitter' (adicionando um ruído entre -0.15 e 0.15)
np.random.seed(42)
jitter = 0.15
x_jitter = x_num + np.random.uniform(-jitter, jitter, size=len(x_num))
y_jitter = y_num + np.random.uniform(-jitter, jitter, size=len(y_num))
z_jitter = z_num + np.random.uniform(-jitter, jitter, size=len(z_num))

# Plotando as classes separadamente para gerar a legenda
for classe, cor, label in zip(['e', 'p'], ['#2ecc71', '#e74c3c'], ['Comestível', 'Venenoso']):
    # Filtra apenas as linhas correspondentes à classe atual
    idx = df['class'] == classe

    # Plota passando o parâmetro 'label'
    ax.scatter(x_jitter[idx], y_jitter[idx], z_jitter[idx],
               c=cor, label=label, alpha=0.5, s=15)

ax.set_xlabel(f'{x_col} (Numérico)')
ax.set_ylabel(f'{y_col} (Numérico)')
ax.set_zlabel(f'{z_col} (Numérico)')
plt.title('Dispersão 3D com Jitter: Odor vs Raiz vs População')

# Adicionando a legenda no canto superior direito
ax.legend(title='Classe', loc='upper right')

plt.show()
```

## 6.1 - Sempre que falamos de Data Science, lembramos dos diferentes tipos de dados. E, junto deles, surgem os temidos dados categóricos!

Esses dados são bem diferentes das variáveis numéricas, então a gente precisa dar uma atenção extra pra trabalhar com eles de forma eficaz. Bora transformar alguns deles usando os métodos que já vimos, como o **OneHotEncoder**! 🔥

Depois disso, vamos mergulhar no **Feature Engineering**!
Essa técnica é super poderosa e serve para extrair informações mais relevantes dos dados que já temos. Ela consiste em selecionar, manipular e transformar colunas e linhas do dataset, dando um novo significado ou enriquecendo as informações. Um exemplo clássico é a criação de uma coluna de IMC (Índice de Massa Corporal) a partir das colunas de altura e peso.

Agora é sua vez! Solte a criatividade e use sua intuição pra criar novos atributos a partir do que já está disponível.

**Dica**: Pegue as colunas que você já tem e use como base pra criar outras com características diferentes.

```python
# FEATURE ENGINEERING

# Removendo a coluna constante (não tem poder preditivo)
df = df.drop('veil-type', axis=1)

# Separando as Características (X) do Alvo (y)
X = df.drop('class', axis=1)
y = df['class']

# Transformando a variável alvo em números binários
# 0 = Comestível (Seguro), 1 = Venenoso (Perigo)
y = y.map({'e': 0, 'p': 1})
```

```python
# ONE-HOT ENCODING:

# O parâmetro dtype=int força o Pandas a usar 0 e 1 em vez de True e False
X_encoded = pd.get_dummies(X, dtype=int)

# Verificação dos resultados:
print("--- Resultados ---")
print(f"Quantidade de características ANTES do One-Hot Encoding:  {X.shape[1]} colunas")
print(f"Quantidade de características DEPOIS do One-Hot Encoding: {X_encoded.shape[1]} colunas\n")

print("Primeiras 5 linhas das características transformadas (X_encoded):")
display(X_encoded.head())

print("\nPrimeiras 5 linhas do alvo transformado (y):")
display(y.head())
```

## 6.2 - Em poucos termos, computadores não conseguem entender texto de maneira alguma, mas números, ah, esses as máquinas entendem bem pra caramba! Então, que tal usar as técnicas que aprendemos pra deixar tudo necessário em números?

**Dica**: Sempre que temos uma feature categórica que possui apenas 2 elementos, podemos transformar ela em uma flag binária! Agora o que isso significa eu deixo pra você pesquisar 😉

```python
# Transformando as colunas categóricas de 2 elementos em Flags Binárias
df['class'] = df['class'].map({'e': 0, 'p': 1})
df['bruises'] = df['bruises'].map({'f': 0, 't': 1})
df['gill-size'] = df['gill-size'].map({'n': 0, 'b': 1})
df['stalk-shape'] = df['stalk-shape'].map({'e': 0, 't': 1})

# Verificando o resultado
display(df[['class', 'bruises', 'gill-attachment', 'gill-spacing', 'gill-size', 'stalk-shape']].head())
```

### Foram transformadas:

- `class`: e → 0, p → 1
- `bruises`: f → 0, t → 1
- `gill-size`: n → 0, b → 1
- `stalk-shape`: e → 0, t → 1
